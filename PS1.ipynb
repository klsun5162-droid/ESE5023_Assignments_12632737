```python
#Matrix multiplication
import random
import numpy as np

M1=np.random.randint(0,51,(5,10))
M2=np.random.randint(0,51, (10,5))

print("The matrix M1 is:\n", M1)
print("The matrix M2 is:\n", M2)

def Matrix_Multip(M1,M2):
    result=np.zeros((len(M1),len(M2[0])),dtype=int)
    for i in range(len(M1)):
        for j in range(len(M2[0])):
            for k in range(len(M2)):
                result[i][j]+=M1[i][k]*M2[k][j]
    print("The result of the matrix multiplication is:\n", result)

Matrix_Multip(M1,M2)
```

```python
#Pascal triangle
def Pascal_triangle(k):
    triangle = []
    for i in range(k):
        row = [1] * (i + 1)
        for j in range(1, i):
            row[j] = triangle[i - 1][j - 1] + triangle[i - 1][j]
        triangle.append(row)
    return row

print(Pascal_triangle(100))
print(Pascal_triangle(200))
```

```python
#5.2
def Find_expression():
    num="123456789"
    import numpy as np
    options=["+","-",""]
    result_num=np.zeros((100,2),dtype=int)

    for target in range(1, 101):
        count = 0
        for i in range(3):
            for j in range(3):
                for k in range(3):
                    for l in range(3):
                        for m in range(3):
                            for n in range(3):
                                for o in range(3):
                                    for p in range(3):
                                        expression=(num[0]+options[i]+num[1]+options[j]+num[2]+options[k]+num[3]+options[l]+num[4]+options[m]+num[5]+options[n]+num[6]+options[o]+num[7]+options[p]+num[8])
                                        if eval(expression) == target:
                                            count += 1
        result_num[target-1,0] = target
        result_num[target-1,1] = count
    print(result_num)

    max_count = max(result_num[:,1])
    min_count = min(result_num[:,1])
    
    max_i = result_num[np.where(result_num[:,1] == max_count), 0]
    min_i = result_num[np.where(result_num[:,1] == min_count), 0]

    print("The maximum number of expressions is:", max_count, "for i:", max_i)
    print("The minimum number of expressions is:", min_count, "for i:", min_i)


Find_expression()
```
